# Codage

```python
cdict = {chr(i): i-65 for i in range(65,91)}
rev_cdict = {i-65 : chr(i) for i in range(65,91)}

def compactify(s):
    es = ""
    for c in s:
        if str.isalpha(c):
            es = es + str.upper(c)
    return es

def encode(s):
    codes = [cdict[l] for l in compactify(s)]
    code = 0
    for n in range(len(codes)):
        code = code + codes[n]*26**n
    return code
```

- Les dictionnaires cdict et rev_cdict realisent la bijection 
$\{A,\cdots,Z\} \simeq \{0,\cong,25\}$
- compactify est l'homomorphisme de monoides (libres) $\pi: ASCII^* \rightarrow \mathcal\{A,\cdots,Z\}^*$ (autrement dit la "purge")
- encode transforme le texte compactifié en une liste d'entiers $c_0, c_1,\cdots,c_n$ puis les encode cette liste selon la formule
  $$ c_0 + c_1 \cdot 26 + c_2 \cdot 26^2 + c_3 \cdot 26^3 + \cdots + c_n \cdot 26^n $$
  (c'est à dire la liste d'entiers est considerée comme un nombre en base 26)
