### Exemples d'utilisation de la méthode `.subs` de substitution symbolique

Symboles "simples"

```python
a,b, c,d = var('a b c d')
((a+a)^2 + b^2).subs({a:c,b:d})
```

Symboles fonctionels

```python
t = var('t')
y = function('y')(t)
F(r,s) = r + s^2
y1 = F(t,y)
y1
```

```python
expr = diff(y,t) + x^2 + log(x)
expr
```

```python
expr.subs({diff(y,t):y1})
```

```python
expr.subs({log(x):y1})
```
